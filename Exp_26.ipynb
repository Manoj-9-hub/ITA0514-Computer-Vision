```python
import cv2
import numpy as np
import os

def reverse_video(input_path: str, output_path: str) -> None:
    """
    Reverses the frames of a video and writes the result to a new file.
    Reads frames sequentially for performance and compatibility.
    """
    cap = cv2.VideoCapture(input_path)
    if not cap.isOpened():
        raise ValueError(f"Could not open video file: {input_path}")

    # Extract video properties
    fps = cap.get(cv2.CAP_PROP_FPS)
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

    # Read all frames into memory (best performance and reliability for short-to-medium videos)
    print("Reading all frames sequentially...")
    frames = []
    while True:
        ret, frame = cap.read()
        if not ret:
            break
        frames.append(frame)

    cap.release()
    total_frames = len(frames)
    print(f"Read {total_frames} frames. Writing in reverse...")

    # Define the codec and create VideoWriter object
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))

    if not out.isOpened():
        raise RuntimeError(f"Could not open VideoWriter for: {output_path}")

    # Write frames in reverse order
    for frame in reversed(frames):
        out.write(frame)

    out.release()
    print(f"Successfully saved reversed video to: {output_path}")

def create_test_video(path: str, num_frames: int = 60, fps: int = 30) -> None:
    """Generates a simple synthetic MP4 video for testing purposes."""
    width, height = 640, 480
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    out = cv2.VideoWriter(path, fourcc, fps, (width, height))

    for i in range(num_frames):
        # Create a black frame
        frame = np.zeros((height, width, 3), dtype=np.uint8)

        # Draw a moving white circle
        x = int(50 + (width - 100) * (i / num_frames))
        y = int(height / 2 + 100 * np.sin(2 * np.pi * i / 30))
        cv2.circle(frame, (x, y), 30, (255, 255, 255), -1)

        # Overlay the frame index
        cv2.putText(frame, f"Frame {i}", (50, 50), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)

        out.write(frame)
    out.release()
    print(f"Synthetic test video created at: {path}")

# --- Create, Run and Verify ---
input_file = "test_input.mp4"
output_file = "test_output_reversed.mp4"

# Create a mock video to test the script
create_test_video(input_file, num_frames=60, fps=30)

# Run the reverse operation
reverse_video(input_file, output_file)
```
